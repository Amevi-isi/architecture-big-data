```python
from pyarrow.fs import HadoopFileSystem

try:
    # Connect to HDFS
    fs = HadoopFileSystem(host='namenode', port=9000)
    
    # Test connection by listing root directory
    contents = fs.get_file_info(['/'])
    print("Connection successful!")
    print(f"Root directory contents: {contents}")
    
    # Additional verification
    print(f"Home directory: {fs.get_file_info(fs.home_directory)}")
    print(f"Used capacity: {fs.get_capacity()} bytes")
    
except Exception as e:
    print(f"Connection failed: {str(e)}")
```
